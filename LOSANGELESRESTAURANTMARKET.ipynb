```python
print("=== NÚMERO MÉDIO DE ASSENTOS POR TIPO ===")
media_assentos_tipo = rest_data_us_upd.groupby('object_type')['number'].agg(['mean', 'count', 'std']).round(2)
media_assentos_tipo = media_assentos_tipo.sort_values('mean', ascending=False)
print(media_assentos_tipo)
print("\n=== RANKING POR MÉDIA DE ASSENTOS ===")
ranking = rest_data_us_upd.groupby('object_type')['number'].mean().sort_values(ascending=False)
for i, (tipo, media) in enumerate(ranking.items(), 1):
    print(f"{i}º {tipo}: {media:.1f} assentos")
```

```python
plt.style.use('default')
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
ax1 = axes[0, 0]
ranking.plot(kind='bar', ax=ax1, color='skyblue', edgecolor='black')
ax1.set_title('Média de Assentos por Tipo de Estabelecimento', fontweight='bold')
ax1.set_xlabel('Tipo de Estabelecimento')
ax1.set_ylabel('Número Médio de Assentos')
ax1.tick_params(axis='x', rotation=45)
ax1.grid(axis='y', alpha=0.3)
ax2 = axes[0, 1]
rest_data_us_upd.boxplot(column='number', by='object_type', ax=ax2)
ax2.set_title('Distribuição de Assentos por Tipo')
ax2.set_xlabel('Tipo de Estabelecimento')
ax2.set_ylabel('Número de Assentos')
plt.suptitle('')
ax3 = axes[1, 0]
ranking.plot(kind='barh', ax=ax3, color='lightcoral', edgecolor='black')
ax3.set_title('Ranking: Média de Assentos por Tipo', fontweight='bold')
ax3.set_xlabel('Número Médio de Assentos')
ax3.set_ylabel('Tipo de Estabelecimento')
ax3.grid(axis='x', alpha=0.3)
ax4 = axes[1, 1]
contagem_tipos = rest_data_us_upd['object_type'].value_counts()
scatter_data = pd.DataFrame({
    'quantidade': contagem_tipos.values,
    'media_assentos': [ranking[tipo] for tipo in contagem_tipos.index],
    'tipo': contagem_tipos.index
})
scatter = ax4.scatter(scatter_data['quantidade'], scatter_data['media_assentos'],
                     s=100, alpha=0.7, c='orange', edgecolors='black')
ax4.set_title('Quantidade vs Média de Assentos por Tipo', fontweight='bold')
ax4.set_xlabel('Quantidade de Estabelecimentos')
ax4.set_ylabel('Média de Assentos')
ax4.grid(True, alpha=0.3)
for i, row in scatter_data.iterrows():
    ax4.annotate(row['tipo'], (row['quantidade'], row['media_assentos']),
                xytext=(5, 5), textcoords='offset points', fontsize=9)
plt.tight_layout()
plt.show()
```

```python
import re

def extrair_nome_rua(endereco):
    """
    Extrai o nome da rua de um endereço completo
    """
    endereco = str(endereco).strip()
    endereco_sem_numero = re.sub(r'^\d+\s*', '', endereco)
    endereco_limpo = re.sub(r'^(N|S|E|W|NE|NW|SE|SW)\s+', '', endereco_sem_numero)
    endereco_final = re.sub(r'\s+\d+.*$', '', endereco_limpo)
    return endereco_final.strip()
exemplos = [
    "3708 N EAGLE ROCK BLVD",
    "1814 W SUNSET BLVD",
    "6801 HOLLYWOOD BLVD 253",
    "100 WORLD WAY 120"
]
for endereco in exemplos:
    nome_rua = extrair_nome_rua(endereco)
    print(f"'{endereco}' -> '{nome_rua}'")
rest_data_us_upd['street_name'] = rest_data_us_upd['address'].apply(extrair_nome_rua)
print(rest_data_us_upd['street_name'].value_counts().head(10))
```

```python
print("=== CONTAGEM DE RESTAURANTES POR RUA ===")
ruas_contagem = rest_data_us_upd['street_name'].value_counts()
print("Total de ruas diferentes:", len(ruas_contagem))
print("\nTop 15 ruas com mais restaurantes:")
print(ruas_contagem.head(15))
top_10_ruas = ruas_contagem.head(10)
print("\n=== TOP 10 RUAS ===")
for i, (rua, quantidade) in enumerate(top_10_ruas.items(), 1):
    print(f"{i}º {rua}: {quantidade} restaurantes")
plt.figure(figsize=(12, 8))
bars = plt.barh(range(len(top_10_ruas)), top_10_ruas.values,
                color='lightcoral', edgecolor='black', alpha=0.8)
plt.yticks(range(len(top_10_ruas)), top_10_ruas.index)
plt.xlabel('Número de Restaurantes', fontsize=12, fontweight='bold')
plt.ylabel('Nome da Rua', fontsize=12, fontweight='bold')
plt.title('Top 10 Ruas com Maior Número de Restaurantes',
          fontsize=14, fontweight='bold', pad=20)
for i, (bar, valor) in enumerate(zip(bars, top_10_ruas.values)):
    plt.text(bar.get_width() + 1, bar.get_y() + bar.get_height()/2,
             str(valor), ha='left', va='center', fontweight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
numero_ruas_um_restaurante = (ruas_contagem == 1).sum()
print(f"Number of streets with only one restaurant: {numero_ruas_um_restaurante}")
```

```python
top_10_ruas = ruas_contagem.head(10)
print("Top 10 ruas:", top_10_ruas.index.tolist())
restaurantes_top_ruas = rest_data_us_upd[rest_data_us_upd['street_name'].isin(top_10_ruas.index)]
print(f"\nTotal de restaurantes nas top 10 ruas: {len(restaurantes_top_ruas)}")
print("=== ESTATÍSTICAS DE ASSENTOS NAS TOP 10 RUAS ===")
print(restaurantes_top_ruas['number'].describe())
print("\n=== COMPARAÇÃO: TOP 10 RUAS vs TODAS AS RUAS ===")
print("Média de assentos nas top 10 ruas:", restaurantes_top_ruas['number'].mean().round(2))
print("Média de assentos em todas as ruas:", rest_data_us_upd['number'].mean().round(2))
```

```python
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.hist(restaurantes_top_ruas['number'], bins=30, alpha=0.7, color='lightcoral', edgecolor='black')
plt.title('Distribuição de Assentos\n(Top 10 Ruas)')
plt.xlabel('Número de Assentos')
plt.ylabel('Frequência')
plt.subplot(1, 3, 2)
plt.hist(rest_data_us_upd['number'], bins=30, alpha=0.7, color='skyblue', edgecolor='black')
plt.title('Distribuição de Assentos\n(Todas as Ruas)')
plt.xlabel('Número de Assentos')
plt.ylabel('Frequência')
plt.subplot(1, 3, 3)
plt.hist(restaurantes_top_ruas['number'], bins=30, alpha=0.6, label='Top 10 ruas', color='lightcoral')
plt.hist(rest_data_us_upd['number'], bins=30, alpha=0.6, label='Todas as ruas', color='skyblue')
plt.title('Comparação das Distribuições')
plt.xlabel('Número de Assentos')
plt.ylabel('Frequência')
plt.legend()
plt.tight_layout()
plt.show()
```

It is evident that the top 10 streets account for a total of 2,399 restaurants, indicating a high concentration of establishments in these areas. Regarding seating capacity, the average is 47.16 seats per restaurant on these streets, compared to an overall average of 43.7 seats across all streets—a difference of just 2.4 seats. This suggests that, despite these being busy streets, the average seating capacity is only slightly higher; the difference is not substantial, reflecting a mix of both large and small restaurants.

When looking at the Los Angeles restaurant market, it is clear that it is vast and highly competitive. There are over 9,000 establishments, the vast majority being traditional restaurants. This demonstrates that, while the sector is robust, the market is fiercely contested; entering this segment would mean competing directly with thousands of well-established businesses. In contrast, the coffee shop segment is much smaller, accounting for just 4.5% of the total. While this might seem small at first glance, it actually represents a significant opportunity. A smaller niche implies less direct competition and more room to offer something unique. For an innovative coffee shop featuring robotic technology, this distinction becomes even more strategic, as the concept launches with a built-in standout factor that grabs attention and sparks curiosity. Regarding size, the average coffee shop in the city has 25 seats; a space of this size would be ideal for your proposal. It is sufficient to serve customers well and generate solid revenue without incurring excessive rent or infrastructure costs. Furthermore, a coffee shop of this scale maintains a comfortable, modern, and welcoming atmosphere—qualities that align perfectly with a tech-forward, distinctive concept. Technology is precisely what sets the business apart. A robotic coffee shop sells more than just coffee; it sells an experience. Automation can lower operating costs, ensure consistent quality, and attract customers through the novelty factor. In a market dominated by independent, traditional establishments, offering something innovative can quickly make the brand stand out. Looking ahead, there is certainly strong potential to turn the business into a chain. The coffee shop model is easier to replicate than a large restaurant because it requires less space and staff, and allows for a high degree of standardization. Starting with a well-located pilot unit—testing the concept and refining processes—is a prudent approach before expanding to other neighborhoods or even other cities. Overall, opening a robotic coffee shop with around 25 seats appears to be the most strategic choice. It is a smart way to enter the market, avoiding the fiercest competition from traditional restaurants and leveraging innovation as a key competitive advantage.
